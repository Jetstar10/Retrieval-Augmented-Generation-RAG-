RAG is a method to combine LLMs with a retrieval systems

The retrieval system can search through vast sources of external information such as documents, databases, knowledge bases whenever the LLM needs additional information to give better answers

RAG system can retrieve the most relevant info from all the documents to give the best answer

Token in the context of language models is a unit of text an LLM processes

Tokens can be as short as one character or as long as one word.

The limit of tokens that an LLM can process at once is called the Context Window

Vector embedding is the mathematical representation of words sentences and images.

Each number in the vector embeddings is called a dimension and each word with similar semantic meaning have their dimensions similar to each other.

Make sure the embelling models used for documents and user query are the same so they understand each other when retrieval.

Make sure the dimensions used for the same models are the same too.
